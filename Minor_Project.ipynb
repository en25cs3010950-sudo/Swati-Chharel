```python

# GroupDNA: Simple WhatsApp Chat Analyzer

from datetime import datetime
import numpy as np

# WhatsApp chat export file
file_path = 'hostel_bois.txt'

# Read file
with open(file_path, 'r', encoding='utf-8') as f:
    lines = f.readlines()

# Data structures
messages = []
sender_counts = {}
date_counts = {}
hour_counts = {}

# Parse WhatsApp Chat
for line in lines:
    line = line.strip()

    if not line:
        continue

    try:
        # Check WhatsApp message format
        if " - " not in line:
            continue

        timestamp_str, rest = line.split(" - ", 1)

        # Convert timestamp
        dt = datetime.strptime(timestamp_str, "%d/%m/%y, %H:%M")

        # Ignore system messages
        if ":" not in rest:
            continue

        sender, text = rest.split(":", 1)

        sender = sender.strip()
        text = text.strip()

        # Skip media & deleted messages
        if text in ("<Media omitted>", "This message was deleted"):
            continue

        messages.append({
            "timestamp": dt,
            "sender": sender,
            "text": text
        })

        # Count sender
        sender_counts[sender] = sender_counts.get(sender, 0) + 1

        # Count date
        date_str = dt.strftime("%d/%m/%Y")
        date_counts[date_str] = date_counts.get(date_str, 0) + 1

        # Count hour
        hour_counts[dt.hour] = hour_counts.get(dt.hour, 0) + 1

    except Exception:
        continue
# Heatmap

participants = sorted(sender_counts.keys())

participant_index = {
    name: i for i, name in enumerate(participants)
}

heatmap = np.zeros((len(participants), 24), dtype=int)

for msg in messages:
    row = participant_index[msg["sender"]]
    col = msg["timestamp"].hour
    heatmap[row][col] += 1

# Report
print("=" * 60)
print("                GROUPDNA REPORT")
print("=" * 60)

print(f"Total Messages      : {len(messages)}")
print(f"Total Participants  : {len(participants)}")

print("-" * 60)
print("MESSAGES PER PERSON")

for person, count in sorted(sender_counts.items(),
                            key=lambda x: x[1],
                            reverse=True):

    percent = (count / len(messages)) * 100 if messages else 0
    print(f"{person:<20} {count:>5} ({percent:.1f}%)")

print("-" * 60)

if date_counts:
    busiest_day = max(date_counts.items(), key=lambda x: x[1])
    print(f"Busiest Day  : {busiest_day[0]} ({busiest_day[1]} messages)")

if hour_counts:
    busiest_hour = max(hour_counts.items(), key=lambda x: x[1])
    print(f"Busiest Hour : {busiest_hour[0]:02d}:00 ({busiest_hour[1]} messages)")

# Most active person
if sender_counts:
    top_person = max(sender_counts.items(), key=lambda x: x[1])
    print(f"Most Active  : {top_person[0]} ({top_person[1]} messages)")

# Average messages/day
if date_counts:
    avg = len(messages) / len(date_counts)
    print(f"Average Messages/Day : {avg:.2f}")

# Longest message
if messages:
    longest = max(messages, key=lambda x: len(x["text"]))
    print("-" * 60)
    print("Longest Message")
    print("Sender :", longest["sender"])
    print("Length :", len(longest["text"]), "characters")

# Top 10 words
word_counts = {}

for msg in messages:
    words = msg["text"].lower().split()

    for word in words:
        word = word.strip(".,!?()[]{}\"'")

        if len(word) < 2:
            continue

        word_counts[word] = word_counts.get(word, 0) + 1

print("-" * 60)
print("TOP 10 WORDS")

for word, count in sorted(word_counts.items(),
                          key=lambda x: x[1],
                          reverse=True)[:10]:
    print(f"{word:<15} {count}")

# Heatmap
print("-" * 60)
print("ACTIVITY HEATMAP")

print("Hour :", end=" ")
for h in range(24):
    print(f"{h:02d}", end=" ")
print()

for i, person in enumerate(participants):
    print(f"{person[:12]:<12}", end=" ")

    for h in range(24):
        value = heatmap[i][h]

        if value == 0:
            ch = "."
        elif value < 5:
            ch = "░"
        elif value < 10:
            ch = "▒"
        else:
            ch = "█"

        print(ch, end="  ")

    print()

print("=" * 60)
print("Analysis Completed Successfully!")
print("=" * 60)
```
